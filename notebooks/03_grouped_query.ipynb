# Grouped-query attention
Where does GQA sit between MHA and MQA?

```python
import os
import sys
from pathlib import Path
root = Path.cwd() if (Path.cwd() / 'src').exists() else Path.cwd().parent
sys.path.insert(0, str(root / 'src'))
import torch
torch.set_num_threads(1)
torch.manual_seed(42)
smoke = os.environ.get('PYTHONIC_SMOKE') == '1'
```

```python
import matplotlib.pyplot as plt
from pythonic.attention import Attention, kv_cache_bytes
heads = [1, 2, 4, 8]
parameters = [sum(p.numel() for p in Attention(64, 8, count).parameters()) for count in heads]
cache = [kv_cache_bytes(4, 1, 2048, count, 8, 2) / 1024 for count in heads]
fig, axes = plt.subplots(1, 2, figsize=(7, 2.5))
axes[0].plot(heads, parameters, 'o-')
axes[0].set(xlabel='KV heads', ylabel='Parameters')
axes[1].plot(heads, cache, 'o-')
axes[1].set(xlabel='KV heads', ylabel='FP16 cache KiB')
fig.tight_layout()
```

GQA trades KV capacity for memory. These counts do not measure model quality.
