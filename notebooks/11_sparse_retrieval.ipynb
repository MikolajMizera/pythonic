# Sparse retrieval
Can a question recover its source abstract?

```python
import os
import sys
from pathlib import Path
root = Path.cwd() if (Path.cwd() / 'src').exists() else Path.cwd().parent
sys.path.insert(0, str(root / 'src'))
import torch
torch.set_num_threads(1)
torch.manual_seed(42)
smoke = os.environ.get('PYTHONIC_SMOKE') == '1'
```

```python
from pythonic.data import fixtures, load_papers, split_papers
from pythonic.retrieval import BM25, retrieval_metrics
papers = fixtures() if smoke else load_papers(root / 'data/pubmedqa.json')
questions = papers if smoke else split_papers(papers).test
index = BM25([section for paper in papers for section in paper.sections])
for hit in index.search(questions[0].question, limit=3):
    print(hit.section.pmid, hit.section.index, round(hit.score, 3), hit.section.text[:100])
print(retrieval_metrics(index, questions))
```

Only abstract sections are indexed. Recall asks whether the source PMID appears among the retrieved sections.
