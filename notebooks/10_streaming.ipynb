# Streaming generation
What does the client receive before generation ends?

```python
import os
import sys
from pathlib import Path
root = Path.cwd() if (Path.cwd() / 'src').exists() else Path.cwd().parent
sys.path.insert(0, str(root / 'src'))
import torch
torch.set_num_threads(1)
torch.manual_seed(42)
smoke = os.environ.get('PYTHONIC_SMOKE') == '1'
```

```python
from fastapi.testclient import TestClient
from pythonic.model import Decoder, DecoderConfig
from pythonic.serving import create_app
model = Decoder(DecoderConfig(width=16, layers=1, context=32))
with TestClient(create_app(model, concurrency=1)) as client:
    with client.stream('POST', '/generate', json={'prompt': 'Trial', 'max_new': 3}) as response:
        for line in response.iter_lines():
            if line:
                print(line)
```

Each event contains a token ID and decoded text. The fixture model is untrained; cancellation returns the capacity slot after any running worker finishes.
