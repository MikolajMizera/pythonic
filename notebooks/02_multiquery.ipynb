# Multi-query attention
How much cache memory does sharing KV heads save?

```python
import os
import sys
from pathlib import Path
root = Path.cwd() if (Path.cwd() / 'src').exists() else Path.cwd().parent
sys.path.insert(0, str(root / 'src'))
import torch
torch.set_num_threads(1)
torch.manual_seed(42)
smoke = os.environ.get('PYTHONIC_SMOKE') == '1'
```

```python
from pythonic.attention import Attention, kv_cache_bytes
for kv_heads in (4, 1):
    model = Attention(32, 4, kv_heads)
    parameters = sum(p.numel() for p in model.parameters())
    cache = kv_cache_bytes(4, 1, 1024, kv_heads, 8)
    print(kv_heads, 'KV heads:', parameters, 'parameters;', cache, 'cache bytes')
    assert model(torch.randn(2, 8, 32)).shape == (2, 8, 32)
```

The cache counts keys and values for every layer. Fewer KV heads reduce it linearly.
