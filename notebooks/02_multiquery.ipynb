# Multi-query attention
What changes when every query head shares keys and values?

```python
import os
import sys
from pathlib import Path
root = Path.cwd() if (Path.cwd() / 'src').exists() else Path.cwd().parent
sys.path.insert(0, str(root / 'src'))
import torch
torch.set_num_threads(1)
torch.manual_seed(42)
smoke = os.environ.get('PYTHONIC_SMOKE') == '1'
```

```python
from pythonic.attention import Attention
for kv_heads in (4, 1):
    model = Attention(32, 4, kv_heads)
    print(kv_heads, 'KV heads:', sum(p.numel() for p in model.parameters()), 'parameters')
    assert model(torch.randn(2, 8, 32)).shape == (2, 8, 32)
```

Sharing KV projections reduces parameters. Query heads remain separate.
