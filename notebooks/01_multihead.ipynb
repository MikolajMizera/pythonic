# Multi-head attention
Can earlier tokens see later ones?

```python
import os
import sys
from pathlib import Path
root = Path.cwd() if (Path.cwd() / 'src').exists() else Path.cwd().parent
sys.path.insert(0, str(root / 'src'))
import torch
torch.set_num_threads(1)
torch.manual_seed(42)
smoke = os.environ.get('PYTHONIC_SMOKE') == '1'
```

```python
from pythonic.attention import Attention
model = Attention(32, 4)
x = torch.randn(1, 12, 32)
y = x.clone()
y[:, 6:] += 10
print('parameters:', sum(p.numel() for p in model.parameters()))
print('earlier output difference:', (model(x)[:, :6] - model(y)[:, :6]).abs().max().item())
```

Causal masking keeps earlier outputs independent of future tokens.
