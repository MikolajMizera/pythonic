# Evidence budget
Which passages fit, and can every quote be traced?

```python
import os
import sys
from pathlib import Path
root = Path.cwd() if (Path.cwd() / 'src').exists() else Path.cwd().parent
sys.path.insert(0, str(root / 'src'))
import torch
torch.set_num_threads(1)
torch.manual_seed(42)
smoke = os.environ.get('PYTHONIC_SMOKE') == '1'
```

```python
from pythonic.context import select_context, validate_citations
from pythonic.data import fixtures
from pythonic.retrieval import BM25
sections = [section for paper in fixtures() for section in paper.sections]
hits = BM25(sections).search('treatment fever')
for budget in (40, 80, 120):
    context = select_context('Does treatment reduce fever?', hits, budget, len)
    validate_citations(context.citations, {(s.pmid, s.index): s for s in sections})
    print(budget, 'character budget:', repr(context.text), context.tokens, 'used')
```

This fixture counts characters to make truncation visible. Real classification uses tokenizer counts; matching a quote proves provenance, not that it supports the decision.
