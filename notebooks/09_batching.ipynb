# Batch scheduling
Can short requests use slots vacated by completed requests?

```python
import os
import sys
from pathlib import Path
root = Path.cwd() if (Path.cwd() / 'src').exists() else Path.cwd().parent
sys.path.insert(0, str(root / 'src'))
import torch
torch.set_num_threads(1)
torch.manual_seed(42)
smoke = os.environ.get('PYTHONIC_SMOKE') == '1'
```

```python
import matplotlib.pyplot as plt
from pythonic.scheduling import Request, simulate
requests = tuple(Request(str(i), 16, count) for i, count in enumerate([40, 2, 3, 4, 2, 5, 1, 3]))
results = [simulate(requests, continuous=mode, max_sequences=2) for mode in (False, True)]
for name, result in zip(['static', 'continuous'], results):
    print(name, round(result.duration, 2), 'time units;', round(result.throughput, 3), 'tokens/unit')
plt.bar(['static', 'continuous'], [result.duration for result in results])
plt.ylabel('Simulated time units')
```

The cost model charges one unit per iteration plus 0.01 per token. These are scheduling comparisons, not measured server throughput.
